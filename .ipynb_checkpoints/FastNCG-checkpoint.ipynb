```julia
using NLPModels
using LinearAlgebra
using Printf
using Statistics
using BenchmarkTools
using FastNCG

import NLPModels: reset!, obj, grad


mutable struct OptResult
    name::String
    time::Float64
    f_final::Float64
    g_final::Float64
    iter::Int
    hvp::Int
    nobj::Int
    ngrad::Int
    x::Vector{Float64}
end


# ============================================================================
# Benchmark settings
# ============================================================================

const BENCH_TOL = 1e-7
const BENCH_MAX_ITER = 1_000
const BENCH_SAMPLES = 10


# ============================================================================
# T-Regression problem
# ============================================================================

function make_problem()
    x0, o, g, hp = make_tregression(200, 500, 0.001)

    label = "T-Regression (dim=200, m=500, nu=0.001)"

    nlp = FastNCG.ClosureNLPModel(
        copy(x0), o, g, hp;
        name="TRegression",
    )

    return label, nlp, copy(x0)
end


# ============================================================================
# FastNCG
# ============================================================================

function benchmark_fastncg(nlp, x0;
                         tol=BENCH_TOL,
                         max_outer=BENCH_MAX_ITER)

    reset!(nlp)

    solver = FastNCG.FastNCGSolver(
        max_outer=max_outer,
    )

    t = @elapsed begin
        x_sol = FastNCG.optimize!(
            solver,
            nlp;
            initial_theta=copy(x0),
            max_iter=max_outer,
            tol=tol,
        )
    end

    # Algorithmic counters before final diagnostic evaluations.
    nobj = nlp.counters.neval_obj
    ngrad = nlp.counters.neval_grad

    f_sol = obj(nlp, x_sol)
    g_sol = norm(grad(nlp, x_sol))

    return OptResult(
        "FastNCG",
        t,
        f_sol,
        g_sol,
        length(solver.loss_history) - 1,
        solver.hvp_count,
        nobj,
        ngrad,
        copy(x_sol),
    )
end


# ============================================================================
# One fresh deterministic run
# ============================================================================

function run_fastncg()
    _, nlp, x0 = make_problem()

    return benchmark_fastncg(
        nlp,
        x0;
        tol=BENCH_TOL,
        max_outer=BENCH_MAX_ITER,
    )
end


# ============================================================================
# Formal benchmark
# ============================================================================

function benchmark_fastncg_formal(; samples=BENCH_SAMPLES)

    # ------------------------------------------------------------------------
    # Warm-up
    # ------------------------------------------------------------------------

    println("Warm-up ...")
    run_fastncg()

    # ------------------------------------------------------------------------
    # Timed runs
    #
    # Construct the deterministic problem outside the timed region.
    # Every repetition receives fresh problem/caches.
    # ------------------------------------------------------------------------

    times = Vector{Float64}(undef, samples)

    for k in 1:samples
        _, nlp, x0 = make_problem()

        solver = FastNCG.FastNCGSolver(
            max_outer=BENCH_MAX_ITER,
        )

        reset!(nlp)

        t = @elapsed begin
            FastNCG.optimize!(
                solver,
                nlp;
                initial_theta=copy(x0),
                max_iter=BENCH_MAX_ITER,
                tol=BENCH_TOL,
            )
        end

        times[k] = t

        @printf(
            "Run %2d/%d: %.6f s\n",
            k, samples, t,
        )
    end

    return (
        median=median(times),
        q25=quantile(times, 0.25),
        q75=quantile(times, 0.75),
        iqr=quantile(times, 0.75) - quantile(times, 0.25),
        minimum=minimum(times),
        maximum=maximum(times),
        times=times,
    )
end


# ============================================================================
# Main
# ============================================================================

function main()

    label, _, _ = make_problem()

    println("="^90)
    println("FastNCG benchmark")
    println("Problem: ", label)
    println("Tolerance: ", BENCH_TOL)
    println("Timed repetitions: ", BENCH_SAMPLES)
    println("="^90)

    # ------------------------------------------------------------------------
    # Deterministic run for algorithmic counters
    # ------------------------------------------------------------------------

    println()
    println("Deterministic run ...")

    r = run_fastncg()

    status =
        r.g_final <= BENCH_TOL ?
        "CONVERGED" :
        "NOT CONVERGED"

    @printf(
        "iter=%d  HVP=%d  Obj=%d  Grad=%d  f=%.8e  |g|=%.4e  [%s]\n",
        r.iter,
        r.hvp,
        r.nobj,
        r.ngrad,
        r.f_final,
        r.g_final,
        status,
    )

    # ------------------------------------------------------------------------
    # Timing
    # ------------------------------------------------------------------------

    println()
    b = benchmark_fastncg_formal()

    println()
    println("="^90)
    println("Timing summary")
    println("="^90)

    @printf("Median : %.6f s\n", b.median)
    @printf("Q25    : %.6f s\n", b.q25)
    @printf("Q75    : %.6f s\n", b.q75)
    @printf("IQR    : %.6f s\n", b.iqr)
    @printf("Min    : %.6f s\n", b.minimum)
    @printf("Max    : %.6f s\n", b.maximum)

    println("="^90)

    return (
        result=r,
        benchmark=b,
        status=status,
    )
end

results = main()
```

```julia
using LinearAlgebra, Printf, NLPModels, BenchmarkTools, Statistics

const TAUS = collect(0.05:0.05:0.95)

function build_problems()
    x0, f, g, hv = make_rosenbrock(100; x0=zeros(100))
    rosen = (name="Rosenbrock", x0=x0, obj=f, grad=g, hprod=hv)

    x0, f, g, hv = make_tregression(100, 200, 0.1)
    treg = (name="TRegression", x0=x0, obj=f, grad=g, hprod=hv)

    return rosen, treg
end

function make_solver(p, τ)
    nlp = FastNCG.ClosureNLPModel(
        copy(p.x0), p.obj, p.grad, p.hprod; name=p.name
    )
    solver = FastNCG.FastNCGSolver(
        sigma_0_ref=1.0, rho=2.0, beta=0.25, gamma=1e-3,
        tau=τ, max_krylov=100, max_outer=10000
    )
    return nlp, solver
end

function solve_once(p, τ)
    nlp, solver = make_solver(p, τ)
    FastNCG.optimize!(
        solver, nlp;
        initial_theta=copy(p.x0), max_iter=10000, tol=1e-7
    )
    return solver
end

function run(p, τ)
    stat_solver = solve_once(p, τ)

    trial = @benchmark FastNCG.optimize!(
        solver, nlp;
        initial_theta=x0,
        max_iter=10000,
        tol=1e-7
    ) setup=begin
        nlp, solver = make_solver($p, $τ)
        x0 = copy($p.x0)
    end evals=1 samples=50 seconds=10

    return (
        hvp  = stat_solver.hvp_count,
        iter = length(stat_solver.loss_history) - 1,
        loss = stat_solver.loss_history[end],
        time = median(trial).time / 1e9
    )
end

function main()
    @printf("%-14s %6s %8s %8s %14s %12s\n",
            "Problem", "tau", "HVP", "Iter", "Loss", "Time(s)")

    results = NamedTuple[]

    for p in build_problems(), τ in TAUS
        r = run(p, τ)

        push!(results, (
            problem=p.name,
            tau=τ,
            r...
        ))

        @printf("%-14s %6.2f %8d %8d %14.6e %12.6f\n",
                p.name, τ, r.hvp, r.iter, r.loss, r.time)
    end

    return results
end

results = main()
```

```julia
using LinearAlgebra
using Printf
using NLPModels

const TAUS = collect(0.05:0.05:0.95)

function build_problems()
    x0, f, g, hv = make_rosenbrock(100; x0=zeros(100))
    rosen = (name="Rosenbrock", x0=x0, obj=f, grad=g, hprod=hv)

    x0, f, g, hv = make_tregression(100, 1000, 0.01)
    treg = (name="TRegression", x0=x0, obj=f, grad=g, hprod=hv)

    return (rosen, treg)
end

function run(problem, tau)
    
    nlp = FastNCG.ClosureNLPModel(
        copy(problem.x0),
        problem.obj,
        problem.grad,
        problem.hprod;
        name=problem.name,
    )

    solver = FastNCG.FastNCGSolver(
        sigma_0_ref=1.0,
        rho=2.0,
        beta=0.25,
        gamma=1e-3,
        tau=tau,
        max_krylov=100,
        max_outer=10000,
    )

    elapsed = @elapsed FastNCG.optimize!(
        solver, nlp;
        initial_theta=copy(problem.x0),
        max_iter=10000,
        tol=1e-7,
    )

    outer = length(solver.loss_history) - 1
    accepted = length(solver.accepted_kappa)
    fallback = outer - accepted

    return (
        hvp=solver.hvp_count,
        outer=outer,
        accepted=accepted,
        fallback=fallback,
        ratio_inc=solver.ratio_increases,
        krylov_ext=solver.krylov_extensions,
        loss=solver.loss_history[end],
        final_g=solver.grad_norm_history[end],
        time=elapsed,
    )
end

function main()
    @printf("%-14s %6s %9s %8s %9s %10s %10s %14s %12s %10s\n",
            "Problem", "tau", "HVP", "Outer", "Accepted",
            "RatioInc", "KrylovExt", "Loss", "Final|g|", "Time(s)")
    println("-"^128)

    results = NamedTuple[]

    for p in build_problems(), tau in TAUS
        r = run(p, tau)

        push!(results, (
            problem=p.name,
            tau=tau,
            hvp=r.hvp,
            outer=r.outer,
            accepted=r.accepted,
#             fallback=r.fallback,
            ratio_inc=r.ratio_inc,
            krylov_ext=r.krylov_ext,
            loss=r.loss,
            final_g=r.final_g,
            time=r.time,
        ))

        @printf("%-14s %6.2f %9d %8d %9d %10d %10d %14.6e %12.3e %10.4f\n",
#             "%-14s %6.2f %9d %8d %9d %9d %10d %10d %14.6e %12.3e %10.4f\n",
                p.name, tau,
                r.hvp,
                r.outer,
                r.accepted,
#                 r.fallback,
                r.ratio_inc,
                r.krylov_ext,
                r.loss,
                r.final_g,
                r.time)
    end

    return results
end

results = main()
```

```julia
module FastNCGStandalone

using LinearAlgebra
using NLPModels
using Printf

import NLPModels: obj, grad!, hprod!

export ClosureNLPModel, FastNCGSolver, optimize!

# ============================================================
# Closure-based NLP model
# ============================================================

mutable struct ClosureNLPModel{F,G,H,D,P}
    meta::NLPModelMeta
    counters::Counters
    obj_fn::F
    grad_fn::G
    hprod_fn::H
    objdiff_fn::D
    precise_diff_fn::P
end

function ClosureNLPModel(x0, obj_fn, grad_fn, hprod_fn;
                         objdiff_fn=nothing, precise_diff_fn=nothing)
    n = length(x0)
    meta = NLPModelMeta(n, x0=copy(x0), name="ClosureNLPModel")
    counters = Counters()
    return ClosureNLPModel(meta, counters, obj_fn, grad_fn, hprod_fn,
                           objdiff_fn, precise_diff_fn)
end

function NLPModels.obj(model::ClosureNLPModel, x)
    return model.obj_fn(x)
end

function NLPModels.grad!(model::ClosureNLPModel, x, g)
    model.grad_fn(g, x)
    return g
end

function NLPModels.hprod!(model::ClosureNLPModel, x, v, Hv; obj_weight=1.0)
    model.hprod_fn(Hv, x, v)

    if obj_weight != 1.0
        rmul!(Hv, obj_weight)
    end

    return Hv
end

# ============================================================
# Solver
# ============================================================

Base.@kwdef mutable struct FastNCGSolver
    sigma_0_ref::Float64 = 1.0
    sigma_min::Float64 = 1e-8
    rho::Float64 = 2.0
    beta::Float64 = 0.25
    gamma::Float64 = 1e-3
    tau::Float64 = 0.1

    max_outer::Int = 500
    max_krylov::Int = 0
    max_inner::Int = 0

    projection_batch::Bool = true

    max_newton::Int = 10
    tol_newton::Float64 = 1e-3
    max_target_retries::Int = 16

    fp_safeguard::Bool = true
    fp_factor::Float64 = 10.0

    verbose::Bool = false

    status::Symbol = :not_started

    hvp_count::Int = 0
    grad_count::Int = 0
    func_count::Int = 0
    krylov_extensions::Int = 0
    ratio_increases::Int = 0
    newton_iters_total::Int = 0
    safeguard_checks::Int = 0

    accepted_kappa::Vector{Float64} = Float64[]
    accepted_ratio_errors::Vector{Float64} = Float64[]
    attained_dims::Vector{Int} = Int[]

    loss_history::Vector{Float64} = Float64[]
    grad_norm_history::Vector{Float64} = Float64[]
    time_history::Vector{Float64} = Float64[]

    total_time::Float64 = 0.0
end

# ============================================================
# Reduced shifted system
# ============================================================

mutable struct TargetWorkspace
    d::Vector{Float64}
    l::Vector{Float64}
    rhs::Vector{Float64}
    u::Vector{Float64}
    w::Vector{Float64}
end

function _target_workspace(n, j)
    return TargetWorkspace(
        zeros(Float64, j),
        zeros(Float64, max(j-1, 0)),
        zeros(Float64, j),
        zeros(Float64, j),
        zeros(Float64, j)
    )
end

function _resize_target_workspace!(ws::TargetWorkspace, j)
    resize!(ws.d, j)
    resize!(ws.l, max(j-1, 0))
    resize!(ws.rhs, j)
    resize!(ws.u, j)
    resize!(ws.w, j)
    return ws
end

function _factor!(ws::TargetWorkspace, diag, off, lambda)
    j = length(diag)
    _resize_target_workspace!(ws, j)

    @inbounds for i in 1:j
        ws.d[i] = diag[i]+lambda
    end

    if j == 0
        return false
    end

    if !(isfinite(ws.d[1]) && ws.d[1] > 0)
        return false
    end

    @inbounds for i in 1:j-1
        ws.l[i] = off[i]/ws.d[i]
        ws.d[i+1] -= ws.l[i]*off[i]

        if !(isfinite(ws.d[i+1]) && ws.d[i+1] > 0)
            return false
        end
    end

    return true
end

function _solve_factored!(x, ws::TargetWorkspace, rhs, off)
    j = length(rhs)

    if j == 0
        return x
    end

    x[1] = rhs[1]

    @inbounds for i in 2:j
        x[i] = rhs[i]-ws.l[i-1]*x[i-1]
    end

    x[j] /= ws.d[j]

    @inbounds for i in j-1:-1:1
        x[i] = (x[i]-off[i]*x[i+1])/ws.d[i]
    end

    return x
end

function _solve_into!(x, ws::TargetWorkspace, diag, off, lambda, rhs)
    _factor!(ws, diag, off, lambda) || return false
    _solve_factored!(x, ws, rhs, off)
    return all(isfinite, x)
end

function _target_eval!(ws::TargetWorkspace, diag, off, gnorm, sigma, lambda)
    j = length(diag)
    _resize_target_workspace!(ws, j)

    fill!(ws.rhs, 0.0)
    ws.rhs[1] = gnorm

    _solve_into!(ws.u, ws, diag, off, lambda, ws.rhs) || return nothing

    unorm = norm(ws.u)

    if !(isfinite(unorm) && unorm > 0)
        return nothing
    end

    F = lambda-sigma*unorm

    copyto!(ws.rhs, ws.u)
    _solve_into!(ws.w, ws, diag, off, lambda, ws.rhs) || return nothing

    derivative = 1.0+sigma*dot(ws.u, ws.w)/unorm

    if !(isfinite(F) && isfinite(derivative) && derivative > 0)
        return nothing
    end

    return F, derivative, unorm
end

# ============================================================
# Spectral lower bound
# ============================================================

function _tridiagonal_minimum(diag, off)
    j = length(diag)

    if j == 1
        return diag[1]
    end

    T = SymTridiagonal(diag, off)
    return eigmin(T)
end

# ============================================================
# Target-ratio solve
# ============================================================

function _target!(s::FastNCGSolver, ws::TargetWorkspace, diag, off, gnorm, sigma)
    j = length(diag)

    if j == 0 || !(isfinite(gnorm) && gnorm > 0) ||
       !(isfinite(sigma) && sigma > 0)
        return nothing
    end

    theta_min = _tridiagonal_minimum(diag, off)

    if !isfinite(theta_min)
        return nothing
    end

    a = max(0.0, -theta_min)

    scale = max(
        1.0,
        abs(theta_min),
        maximum(abs, diag),
        isempty(off) ? 0.0 : maximum(abs, off)
    )

    gap = max(
        100.0*eps(Float64)*scale,
        eps(Float64)*max(1.0, a)
    )

    lambda = max(
        a+gap,
        sqrt(max(sigma*gnorm, eps(Float64)))
    )

    left = a+gap
    left_eval = _target_eval!(ws, diag, off, gnorm, sigma, left)

    retries = 0

    while (left_eval === nothing || left_eval[1] >= 0) &&
          retries < s.max_target_retries

        gap *= 0.1
        left = a+gap
        left_eval = _target_eval!(ws, diag, off, gnorm, sigma, left)
        retries += 1
    end

    if left_eval === nothing || left_eval[1] >= 0
        return nothing
    end

    if lambda <= left
        lambda = 2.0*left
    end

    current = _target_eval!(ws, diag, off, gnorm, sigma, lambda)

    retries = 0

    while current === nothing && retries < s.max_target_retries
        lambda = max(2.0*lambda, left+gap)
        current = _target_eval!(ws, diag, off, gnorm, sigma, lambda)
        retries += 1
    end

    current === nothing && return nothing

    for _ in 1:s.max_newton
        F, derivative, _ = current

        correction = F/derivative

        if abs(correction) <= s.tol_newton*max(1.0, abs(lambda))
            fill!(ws.rhs, 0.0)
            ws.rhs[1] = gnorm

            _solve_into!(ws.u, ws, diag, off, lambda, ws.rhs) ||
                return nothing

            s.newton_iters_total += 1

            return lambda, ws.u
        end

        candidate = lambda-correction

        if !(isfinite(candidate) && candidate > left)
            candidate = 0.5*(lambda+left)
        end

        candidate_eval = _target_eval!(
            ws, diag, off, gnorm, sigma, candidate
        )

        retries = 0

        while candidate_eval === nothing &&
              retries < s.max_target_retries

            candidate = 0.5*(candidate+lambda)

            if candidate <= left
                candidate = 0.5*(lambda+left)
            end

            candidate_eval = _target_eval!(
                ws, diag, off, gnorm, sigma, candidate
            )

            retries += 1
        end

        candidate_eval === nothing && return nothing

        lambda = candidate
        current = candidate_eval
        s.newton_iters_total += 1
    end

    fill!(ws.rhs, 0.0)
    ws.rhs[1] = gnorm

    _solve_into!(ws.u, ws, diag, off, lambda, ws.rhs) ||
        return nothing

    return lambda, ws.u
end

# ============================================================
# Lanczos extension
# ============================================================

function _lanczos!(s::FastNCGSolver, model, x, Q, diag, off, q_prev, q, beta_prev, Hv)
    hprod!(model, x, q, Hv)
    s.hvp_count += 1

    w = copy(Hv)

    if q_prev !== nothing
        @. w -= beta_prev*q_prev
    end

    alpha = dot(q, w)
    @. w -= alpha*q

    correction = dot(q, w)
    @. w -= correction*q
    alpha += correction

    beta_new = norm(w)

    push!(diag, alpha)

    if q_prev !== nothing
        push!(off, beta_prev)
    end

    Q = hcat(Q, q)
    s.krylov_extensions += 1

    threshold = 10.0*eps(Float64)*max(
        norm(Hv),
        abs(alpha),
        abs(beta_prev),
        1.0
    )

    if !(isfinite(beta_new) && beta_new > threshold)
        return Q, diag, off, q, nothing, 0.0, true
    end

    q_new = w/beta_new

    return Q, diag, off, q, q_new, beta_new, false
end

# ============================================================
# Projection criterion
# ============================================================

function _projection(Q, g)
    ng = norm(g)

    if ng == 0
        return 1.0
    end

    return norm(transpose(Q)*g)/ng
end

# ============================================================
# Quadratic finite-precision safeguard
# ============================================================

function _quadratic_descent_margin(diag, off, y, gnorm, beta, lambda, vnorm2)
    j = length(y)

    gTv = -gnorm*y[1]

    vTHv = 0.0

    @inbounds for i in 1:j
        vTHv += diag[i]*y[i]^2
    end

    @inbounds for i in 1:j-1
        vTHv += 2.0*off[i]*y[i]*y[i+1]
    end

    return gTv+0.5*vTHv+beta*lambda*vnorm2
end

function _decrease(s::FastNCGSolver, model, x, trial, v, fx, ft,
                   lambda, diag, off, y, gnorm)

    vnorm2 = sum(abs2, v)
    req = s.beta*lambda*vnorm2

    # Preferred path: model-specific stable objective difference.
    if model.objdiff_fn !== nothing
        delta, _ = model.objdiff_fn(x, trial)
        margin = delta+req
        return isfinite(margin) ? margin <= 0 : nothing
    end

    # Generic direct objective-difference path.
    resolution = s.fp_factor*eps(Float64)*max(
        1.0,
        abs(fx),
        abs(ft)
    )

    if !s.fp_safeguard || req > resolution
        margin = (ft-fx)+req
        return isfinite(margin) ? margin <= 0 : nothing
    end

    # Finite-precision fallback.
    s.safeguard_checks += 1

    margin = _quadratic_descent_margin(
        diag, off, y, gnorm,
        s.beta, lambda, vnorm2
    )

    return isfinite(margin) ? margin <= 0 : nothing
end

# ============================================================
# Reset solver statistics
# ============================================================

function _reset!(s::FastNCGSolver)
    s.status = :not_started

    s.hvp_count = 0
    s.grad_count = 0
    s.func_count = 0
    s.krylov_extensions = 0
    s.ratio_increases = 0
    s.newton_iters_total = 0
    s.safeguard_checks = 0

    empty!(s.accepted_kappa)
    empty!(s.accepted_ratio_errors)
    empty!(s.attained_dims)

    empty!(s.loss_history)
    empty!(s.grad_norm_history)
    empty!(s.time_history)

    s.total_time = 0.0

    return s
end

# ============================================================
# Main optimization routine
# ============================================================

function optimize!(s::FastNCGSolver, model::ClosureNLPModel;
                   initial_theta=nothing, tol=1e-7)

    _reset!(s)

    start_time = time()

    x = initial_theta === nothing ?
        copy(model.meta.x0) :
        copy(initial_theta)

    n = length(x)

    max_krylov = s.max_krylov <= 0 ?
                 n :
                 min(s.max_krylov, n)

    max_inner = s.max_inner <= 0 ?
                3*max_krylov :
                s.max_inner

    g = zeros(Float64, n)
    gt = zeros(Float64, n)
    Hv = zeros(Float64, n)

    fx = obj(model, x)
    s.func_count += 1

    grad!(model, x, g)
    s.grad_count += 1

    gnorm = norm(g)

    push!(s.loss_history, fx)
    push!(s.grad_norm_history, gnorm)
    push!(s.time_history, 0.0)

    if s.verbose
        @printf(
            "%5s %15s %15s %8s %12s %12s\n",
            "iter", "f", "|g|", "j", "sigma", "lambda"
        )

        @printf(
            "%5d %15.7e %15.7e %8s %12s %12s\n",
            0, fx, gnorm, "-", "-", "-"
        )
    end

    if gnorm <= tol
        s.status = :first_order
        s.total_time = time()-start_time
        return x
    end

    sigma_ref = max(s.sigma_min, s.sigma_0_ref)

    ws = _target_workspace(n, 1)

    for outer in 1:s.max_outer
        gnorm = norm(g)

        if gnorm <= tol
            s.status = :first_order
            break
        end

        sigma = sigma_ref

        Q = Matrix{Float64}(undef, n, 0)
        diag = Float64[]
        off = Float64[]

        q_prev = nothing
        q = -g/gnorm
        beta_prev = 0.0

        Q, diag, off, q_prev, q, beta_prev, breakdown =
            _lanczos!(
                s, model, x, Q, diag, off,
                q_prev, q, beta_prev, Hv
            )

        accepted = false
        inner = 0

        while !accepted && inner < max_inner
            inner += 1

            j = length(diag)

            if j == 0
                s.status = :inner_failure
                break
            end

            target = _target!(
                s, ws, diag, off,
                gnorm, sigma
            )

            if target === nothing
                if !breakdown && q !== nothing && j < max_krylov
                    Q, diag, off, q_prev, q, beta_prev, breakdown =
                        _lanczos!(
                            s, model, x, Q, diag, off,
                            q_prev, q, beta_prev, Hv
                        )

                    continue
                end

                sigma *= s.rho
                s.ratio_increases += 1
                continue
            end

            lambda, ybuf = target
            y = copy(ybuf)

            v = Q*y
            vnorm = norm(v)

            if !(isfinite(vnorm) && vnorm > 0)
                sigma *= s.rho
                s.ratio_increases += 1
                continue
            end

            trial = x+v

            if trial == x
                s.status = :numerical_stagnation
                break
            end

            ft = obj(model, trial)
            s.func_count += 1

            grad!(model, trial, gt)
            s.grad_count += 1

            gt_norm = norm(gt)

            if !(isfinite(ft) && isfinite(gt_norm))
                sigma *= s.rho
                s.ratio_increases += 1
                continue
            end

            projection_ratio = _projection(Q, gt)

            projection_ok =
                projection_ratio >= s.tau

            gradient_ok =
                lambda*vnorm >= s.gamma*gt_norm

            decrease_ok = false

            if projection_ok && gradient_ok
                dec = _decrease(
                    s, model,
                    x, trial, v,
                    fx, ft,
                    lambda,
                    diag, off, y,
                    gnorm
                )

                decrease_ok = dec === true
            end
            

            if projection_ok && gradient_ok && decrease_ok
                kappa = lambda/vnorm
                ratio_error = abs(kappa/sigma-1.0)

                x = trial
                fx = ft
                copyto!(g, gt)
                gnorm = gt_norm

                sigma_ref = max(
                    s.sigma_min,
                    kappa/s.rho
                )

                push!(s.accepted_kappa, kappa)
                push!(s.accepted_ratio_errors, ratio_error)
                push!(s.attained_dims, j)

                push!(s.loss_history, fx)
                push!(s.grad_norm_history, gnorm)
                push!(
                    s.time_history,
                    time()-start_time
                )

                accepted = true

                if s.verbose
                    @printf(
                        "%5d %15.7e %15.7e %8d %12.4e %12.4e\n",
                        outer, fx, gnorm,
                        j, sigma, lambda
                    )
                end

                break
            end

            if !projection_ok &&
               !breakdown &&
               q !== nothing &&
               j < max_krylov

                batch = 1

                if s.projection_batch
                    ratio = projection_ratio/max(s.tau, eps(Float64))

                    if ratio < 0.25
                        batch = 5
                    elseif ratio < 0.5
                        batch = 3
                    end
                end

                batch = min(batch, max_krylov-j)

                for _ in 1:batch
                    Q, diag, off, q_prev, q, beta_prev, breakdown =
                        _lanczos!(
                            s, model, x, Q, diag, off,
                            q_prev, q, beta_prev, Hv
                        )

                    if breakdown || q === nothing
                        break
                    end
                end

                continue
            end

            sigma *= s.rho
            s.ratio_increases += 1
        end

        if !accepted
            if s.status == :not_started
                s.status = :inner_failure
            end
            break
        end

        if gnorm <= tol
            s.status = :first_order
            break
        end
    end

    if s.status == :not_started
        if norm(g) <= tol
            s.status = :first_order
        else
            s.status = :max_outer
        end
    end

    s.total_time = time()-start_time

    return x
end

end
```

```julia
using LinearAlgebra
using Random
using Printf
using NLPModels

S = FastNCGStandalone
BLAS.set_num_threads(1)

# 1. 构造 T-regression 问题
n = 200
m = 500
nu = 0.001

x0, f, g!, hv!, objdiff = make_tregression(
    n, m, nu;
    seed=42
)

model = S.ClosureNLPModel(
    x0, f, g!, hv!;
    objdiff_fn=objdiff
)

solver = S.FastNCGSolver(
    max_outer=10000,
#     max_krylov=30,
    verbose=true
)

x = Base.invokelatest(
    () -> S.optimize!(
        solver,
        model;
        initial_theta=copy(x0),
        tol=1e-7
    )
)

(
    status=solver.status,
    loss=solver.loss_history[end],
    gradient_norm=solver.grad_norm_history[end],
    iterations=length(solver.loss_history)-1,
    HVP=solver.hvp_count,
    function_evaluations=solver.func_count,
    gradient_evaluations=solver.grad_count,
    krylov_extensions=solver.krylov_extensions,
    sigma_increases=solver.ratio_increases,
    safeguard_checks=solver.safeguard_checks,
    mean_krylov_dimension=isempty(solver.attained_dims) ? NaN : sum(solver.attained_dims)/length(solver.attained_dims),
    max_krylov_dimension=isempty(solver.attained_dims) ? 0 : maximum(solver.attained_dims),
    seconds=solver.total_time
)
```

```julia
using Printf, Statistics, BenchmarkTools

S = FastNCGStandalone

function compare_tau(A, b, nu, x0;
                     taus=collect(0.05:0.05:0.95),
                     max_outer=1000,
                     tol=1e-7, samples=50, seconds=10.0)

    f(x) = sum(log1p.((A*x-b).^2 ./ nu))

    function g!(out, x)
        r = A*x-b
        mul!(out, A', 2 .* r ./ (nu .+ r.^2))
        out
    end

    function hv!(out, x, v)
        r = A*x-b
        w = 2 .* (nu .- r.^2) ./ (nu .+ r.^2).^2
        mul!(out, A', w .* (A*v))
        out
    end

    function make_run(tau)
        d, p = S.treg_difference_callbacks(A, b, nu)
        model = S.ClosureNLPModel(
            copy(x0), f, g!, hv!;
            objdiff_fn=d, precise_diff_fn=p
        )
        solver = S.FastNCGSolver(
            tau=Float64(tau), gamma=1e-3, max_outer=max_outer
        )
        model, solver
    end

    function solve_once(tau)
        model, solver = make_run(tau)
        S.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        solver
    end

    function bench_once(tau)
        model, solver = make_run(tau)
        S.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        nothing
    end

    println("Warm-up...")
    solve_once(first(taus))

    results = NamedTuple[]

    @printf("%5s %-18s %6s %7s %9s %12s %11s %7s %7s\n",
            "tau", "status", "iter", "HVP", "time(s)",
            "loss", "|g|", "mean-j", "max-j")

    for tau in taus
        solver = solve_once(tau)

        bench = @benchmarkable $bench_once($tau) evals=1
        trial = BenchmarkTools.run(
            bench; samples=samples, seconds=seconds
        )
        t = median(trial).time / 1e9

        row = (
            tau=Float64(tau),
            status=solver.status,
            iterations=length(solver.loss_history)-1,
            HVP=solver.hvp_count,
            seconds=t,
            loss=solver.loss_history[end],
            gradient_norm=solver.grad_norm_history[end],
            mean_dimension=isempty(solver.attained_dims) ?
                           NaN : mean(solver.attained_dims),
            max_dimension=isempty(solver.attained_dims) ?
                          0 : maximum(solver.attained_dims),
            krylov_extensions=solver.krylov_extensions,
            sigma_increases=solver.ratio_increases,
            high_precision_checks=solver.precise_checks,
        )

        push!(results, row)

        @printf("%5.2f %-18s %6d %7d %9.5f %12.5f %11.3e %7.2f %7d\n",
                row.tau, string(row.status), row.iterations, row.HVP,
                row.seconds, row.loss, row.gradient_norm,
                row.mean_dimension, row.max_dimension)
    end

    results
end

results = Base.invokelatest(() -> compare_tau(
    A, b, nu, x0;
#     max_krylov=30,
    samples=50,
    seconds=10.0,
))

```

```julia
using Printf, Statistics, BenchmarkTools

Snew = FastNCGStandalone
Sold = FastNCG

function compare_tau_old_new(A, b, nu, x0;
                             taus=collect(0.05:0.05:0.95),
                             max_outer=1000,
                             tol=1e-7,
                             samples=50,
                             seconds=10.0)

    f(x) = sum(log1p.((A*x-b).^2 ./ nu))

    function g!(out, x)
        r = A*x-b
        mul!(out, A', 2 .* r ./ (nu .+ r.^2))
        out
    end

    function hv!(out, x, v)
        r = A*x-b
        w = 2 .* (nu .- r.^2) ./ (nu .+ r.^2).^2
        mul!(out, A', w .* (A*v))
        out
    end

    # ------------------------------------------------------------
    # New implementation
    # ------------------------------------------------------------

    function make_new(tau)
        d, p = Snew.treg_difference_callbacks(A, b, nu)

        model = Snew.ClosureNLPModel(
            copy(x0), f, g!, hv!;
            objdiff_fn=d,
            precise_diff_fn=p
        )

        solver = Snew.FastNCGSolver(
            tau=Float64(tau),
            gamma=1e-3,
            max_outer=max_outer
        )

        model, solver
    end

    function solve_new(tau)
        model, solver = make_new(tau)
        Snew.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        solver
    end

    function bench_new(tau)
        model, solver = make_new(tau)
        Snew.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        nothing
    end

    # ------------------------------------------------------------
    # Old implementation
    # ------------------------------------------------------------

    function make_old(tau)
        model = Sold.ClosureNLPModel(
            copy(x0), f, g!, hv!
        )

        solver = Sold.FastNCGSolver(
            tau=Float64(tau),
            gamma=1e-3,
            max_outer=max_outer
        )

        model, solver
    end

    function solve_old(tau)
        model, solver = make_old(tau)
        Sold.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        solver
    end

    function bench_old(tau)
        model, solver = make_old(tau)
        Sold.optimize!(solver, model; initial_theta=copy(x0), tol=tol)
        nothing
    end

    # ------------------------------------------------------------
    # Extract common statistics
    # ------------------------------------------------------------

    function summarize(solver, tau, runtime)
        final_g = solver.grad_norm_history[end]
        status = final_g <= tol ? :first_order : :not_converged

        (
            tau=Float64(tau),
            status=status,
            iterations=length(solver.loss_history)-1,
            HVP=solver.hvp_count,
            seconds=runtime,
            loss=solver.loss_history[end],
            gradient_norm=final_g,
            mean_dimension=isempty(solver.attained_dims) ?
                           NaN : mean(solver.attained_dims),
            max_dimension=isempty(solver.attained_dims) ?
                          0 : maximum(solver.attained_dims),
            krylov_extensions=solver.krylov_extensions,
            sigma_increases=solver.ratio_increases,
        )
    end

    # ------------------------------------------------------------
    # Warm-up
    # ------------------------------------------------------------

    println("Warm-up...")
    solve_old(first(taus))
    solve_new(first(taus))

    old_results = NamedTuple[]
    new_results = NamedTuple[]

    # ------------------------------------------------------------
    # Old
    # ------------------------------------------------------------

    println()
    println("="^120)
    println("OLD IMPLEMENTATION")
    println("="^120)

    @printf("%5s %-15s %6s %7s %9s %12s %11s %7s %7s\n",
            "tau", "status", "iter", "HVP", "time(s)",
            "loss", "|g|", "mean-j", "max-j")

    for tau in taus
        solver = solve_old(tau)

        bench = @benchmarkable $bench_old($tau) evals=1
        trial = BenchmarkTools.run(bench; samples=samples, seconds=seconds)
        t = median(trial).time/1e9

        row = summarize(solver, tau, t)
        push!(old_results, row)

        @printf("%5.2f %-15s %6d %7d %9.5f %12.5f %11.3e %7.2f %7d\n",
                row.tau, string(row.status), row.iterations, row.HVP,
                row.seconds, row.loss, row.gradient_norm,
                row.mean_dimension, row.max_dimension)
    end

    # ------------------------------------------------------------
    # New
    # ------------------------------------------------------------

    println()
    println("="^120)
    println("NEW IMPLEMENTATION")
    println("="^120)

    @printf("%5s %-15s %6s %7s %9s %12s %11s %7s %7s\n",
            "tau", "status", "iter", "HVP", "time(s)",
            "loss", "|g|", "mean-j", "max-j")

    for tau in taus
        solver = solve_new(tau)

        bench = @benchmarkable $bench_new($tau) evals=1
        trial = BenchmarkTools.run(bench; samples=samples, seconds=seconds)
        t = median(trial).time/1e9

        row = summarize(solver, tau, t)
        push!(new_results, row)

        @printf("%5.2f %-15s %6d %7d %9.5f %12.5f %11.3e %7.2f %7d\n",
                row.tau, string(row.status), row.iterations, row.HVP,
                row.seconds, row.loss, row.gradient_norm,
                row.mean_dimension, row.max_dimension)
    end

    # ------------------------------------------------------------
    # Direct comparison
    # ------------------------------------------------------------

    println()
    println("="^140)
    println("OLD vs NEW")
    println("="^140)

    @printf("%5s %8s %8s %9s %9s %9s %9s %9s %12s %12s\n",
            "tau", "old-it", "new-it",
            "old-HVP", "new-HVP",
            "old-time", "new-time", "speedup",
            "old-loss", "new-loss")

    for (old, new) in zip(old_results, new_results)
        speedup = old.seconds/new.seconds

        @printf("%5.2f %8d %8d %9d %9d %9.5f %9.5f %9.3f %12.5f %12.5f\n",
                old.tau,
                old.iterations, new.iterations,
                old.HVP, new.HVP,
                old.seconds, new.seconds,
                speedup,
                old.loss, new.loss)
    end

    return (
        old=old_results,
        new=new_results
    )
end

results = Base.invokelatest(() -> compare_tau_old_new(
    A, b, nu, x0;
    samples=50,
    seconds=10.0,
))
```
